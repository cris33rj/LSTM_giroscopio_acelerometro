```python
import os
import pandas as pd

def concatenate_subject_data(base_path, output_file="concatenated_data_forth_trace.csv"):
    """
    Concatenate CSV files from a dataset organized by subject folders, adding a subject_id column.
    
    Args:
        base_path (str): The base directory containing subject folders (e.g., "FORTH_TRACE_DATASET-master").
        output_file (str): The name of the output CSV file. Default is "concatenated_data.csv".
    
    Returns:
        None: Saves the concatenated data to the specified output file.
    """
    # Define the final header for the concatenated DataFrame
    final_columns = [
        "subject_id",
        "device_id",
        "accelerometer_x",
        "accelerometer_y",
        "accelerometer_z",
        "gyroscope_x",
        "gyroscope_y",
        "gyroscope_z",
        "magnetometer_x",
        "magnetometer_y",
        "magnetometer_z",
        "timestamp",
        "activity_label"
    ]

    # Initialize an empty DataFrame to store the concatenated results
    all_data = pd.DataFrame()

    # Loop through each subject's folder
    for subject_id in range(15):  # X goes from 0 to 14
        subject_folder = os.path.join(base_path, f"part{subject_id}")
        if not os.path.exists(subject_folder):
            continue  # Skip if the folder doesn't exist

        # Process each CSV file in the folder
        for device_id in range(0, 6):  # dev1 to dev5
            csv_file = os.path.join(subject_folder, f"part{subject_id}dev{device_id}.csv")
            if not os.path.exists(csv_file):
                continue  # Skip if the file doesn't exist

            # Load the CSV file
            data = pd.read_csv(csv_file,header=None)

            # Insert the subject_id and device_id columns
            data.insert(0, "", subject_id)  # Add device_id
            
            print(f"subject_id is =====> {subject_id}")
            #print(data.columns)
            #print(data)

            # Append to the main DataFrame
            all_data = pd.concat([all_data, data], ignore_index=True, axis=0)
            #print(all_data)

    # Assign final columns to ensure correct order
    all_data.columns = final_columns   
    

    # Save the concatenated data to a CSV file
    all_data.to_csv(output_file, index=False)
    print(f"Data has been successfully concatenated and saved to {output_file}")
    
    return all_data

   

# Example usage:
# concatenate_subject_data("FORTH_TRACE_DATASET-master", "output_file.csv")
```
